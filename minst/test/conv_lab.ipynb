```python
class convolve(Node):

    def __init__(self, data, kernel):
        super().__init__([data, kernel])

    #for simplicity,we transform the 3d data into 2d by flattening the channel dimension
    def im2col(self, data, kernel_size):
        c,w,h= data.shape
        k_w,k_h= kernel_size
        col=np.zeros((c*k_w*k_h,w-k_w+1,h-k_h+1))
        for i in range(k_w):
            for j in range(k_h):
                col[i*k_h+j]=data[:,i:i+w-k_w+1,j:j+h-k_h+1].reshape(c,-1)
        return col

    def compute_value(self):
        data=self.parents[0].value
        kernel=self.parents[1].value
        # limit dimension of data is 3,and kernel is 2
        assert len(data.shape) ==3 and len(kernel.shape) == 2, "data should be 3D and kernel should be 2D"  
        c,w,h= data.shape
        data=self.transform(data)
```
